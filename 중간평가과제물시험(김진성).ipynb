### 문제 2. 순차 구조 프로그램 (15점)

사용자로부터 초 단위의 시간을 입력받아, 시간·분·초로 변환하여 출력하는 프로그램을 작성하시오.

- `//` (정수 나눗셈)과 `%` (나머지) 연산자를 활용
- f-문자열로 결과 출력

```
[출력 예시]  (3723 입력 시)
입력하신 3723초는 1시간 2분 3초 입니다.
```

(참고: 챕터 4 - 구조적 프로그래밍 패러다임)

**[답안 2]** 밑의 code block을 수정하고, 실행하십시오.

````python
# 초 단위 입력
total_sec = int(input("시간을 입력하세요(초): "))

# 시간, 분, 초 계산
hours = total_sec // 3600 # 여기에 시간을 계산하는 코드를 작성하세요. (힌트: 3600으로 나눈 몫)
minutes = (total_sec % 3600) // 60 # 여기에 분을 계산하는 코드를 작성하세요. (힌트: 시간을 제외한 나머지를 60으로 나눈 몫)
seconds = total_sec % 60 # 여기에 남은 초를 계산하는 코드를 작성하세요. (힌트: 60으로 나눈 나머지)

# 결과 출력
print(f"입력하신 {total_sec}초는 {hours}시간 {minutes}분 {seconds}초 입니다.")
````

사용자로부터 점수(0~100)를 입력받아, 성적 기준표에 따라 등급을 출력하는 프로그램을 작성하시오.

| 점수 범위 | 등급 |
|-----------|------|
| 90 ~ 100 | A |
| 80 ~ 89 | B |
| 70 ~ 79 | C |
| 60 ~ 69 | D |

- 점수가 0 미만이거나 100 초과인 경우 "잘못된 입력"을 출력

**[답안 3]** 밑의 code block을 수정하고, 실행하십시오.

````python
# 점수 입력
score = int(input("점수를 입력하세요: "))

# 등급 판별
if score < 0 or score > 100:
    grade = "잘못된 입력"
elif score >= 90:
    grade = "A"
elif score >= 80:
# 여기에 elif/else를 사용하여 B, C, D, F 등급을 판별하는 코드를 작성하세요.
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"


# 결과 출력
print(f"{score}점의 등급은 {grade} 입니다.")
````

### 문제 4. 반복 구조 (15점)

1부터 사용자가 입력한 정수 n까지의 숫자 중 **3의 배수를 제외한** 모든 정수의 합을 구하는 프로그램을 작성하시오.

- for문과 range() 함수 사용
- continue를 사용하여 3의 배수를 건너뛸 것

```
[출력 예시]  (10 입력 시)
3의 배수를 제외한 1부터 10까지의 합은 37입니다.
```

(참고: 챕터 6 - 반복 구조)

**[답안 4]** 밑의 code block을 수정하고, 실행하십시오.

````python
# 정수 입력
n = int(input("정수를 입력하세요: "))

total = 0

# 합계 계산
for i in range(1, n+1):  # 여기를 수정하세요
    if i % 3 == 0:  # 여기에 3의 배수를 건너뛰는 조건을 작성하세요
        continue
    total += i

# 결과 출력
print(f"3의 배수를 제외한 1부터 {n}까지의 합은 {total}입니다.")
````

### 문제 5. 함수 기본 (10점)

두 수와 연산자(+, -, *, /)를 매개변수로 받아 계산 결과를 반환하는 함수 `calculator`를 작성하시오.

(참고: 챕터 7 - 함수)

**[답안 5]** 밑의 code block을 수정하고, 실행하십시오.

````python
def calculator(num1, num2, operator):
    if operator == '+':
       return num1 + num2
    elif operator == '-':
       return num1 - num2
    elif operator == '*':
        return num1 * num2
    elif operator == '/':
        if num2 == 0:  # 0으로 나누는 경우
            return "0으로 나눌 수 없습니다."
        return num1 / num2
    else:
        return "유효하지 않은 연산자입니다."

# 사용자 입력
num1 = float(input("첫 번째 숫자를 입력하세요: "))
num2 = float(input("두 번째 숫자를 입력하세요: "))
operator = input("연산자를 입력하세요 (+, -, *, /): ")

# 함수 호출 및 결과 출력
result = calculator(num1, num2, operator)
print(f"결과: {result}")
````
